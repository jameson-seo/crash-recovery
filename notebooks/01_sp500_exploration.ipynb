```python
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
sp500 = yf.download("^GSPC", "1990-01-01", "2026-10-18")
```

```python
sp500.columns
```

```python
sp500.columns = sp500.columns.droplevel("Ticker")
sp500.columns
```

```python
sp500.tail(10)
```

```python
sp500.shape
```

```python
sp500["peak"] = sp500["Close"].cummax()
sp500["drawdown"] = sp500["Close"] / sp500["peak"] - 1
```

```python
sp500["drawdown"].tail(10)
sp500["drawdown"].min()
```

```python
sp500["drawdown"].idxmin()
```

```python
sp500["drawdown"].tail(10)
```

```python
sp500["in_crash"] = sp500["drawdown"] <= -.2
```

```python
sp500["in_crash"].sum()
```
